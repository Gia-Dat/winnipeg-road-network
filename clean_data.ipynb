```python
import pandas as pd
```

## Address

```python
address = pd.read_csv("data/raw/Addresses_20260919.csv")
address.head(5)
```

```python
# Drop unuseful cols
cols_to_drop = [
    "Street Direction",
    "Street Number Suffix",
    "Unit Type",
    "Unit Number",
    "Point",
    "Ward as of September 17",
    "Full Address",
    "Street Address",
]

address = address.drop(columns=cols_to_drop, errors="ignore")
print(address.shape)
address.head(5)
```

```python
address.isna().sum()
```

```python
# Replace all missing data to None
text_columns = ["Street Type", "Neighbourhood", "Ward", "School Division", "Plow Zone"]

for col in text_columns:
    address[col] = address[col].replace({"": None, "NaN": None})

address.head(5)
```

```python
# Standardize column names
address.columns = [col.strip().lower().replace(" ", "_") for col in address.columns]

# Strip leading/trailing whitespaces
str_cols = address.select_dtypes(include="object").columns
for col in str_cols:
    address[col] = address[col].apply(lambda x: x.strip() if isinstance(x, str) else x)
    
address.head(5)
```

```python
# Extract latitude and longitude
geo_split = address["geometry"].str.extract(r"\((?P<latitude>[^,]+),\s*(?P<longitude>[^)]+)\)")

# Convert to float
address["latitude"] = pd.to_numeric(geo_split["latitude"], errors="coerce")
address["longitude"] = pd.to_numeric(geo_split["longitude"], errors="coerce")

# Drop geometry col
address = address.drop(columns=["geometry"])
address.head(5)
```

```python
# Check duplication
address.duplicated().sum()
```

```python
address.dtypes
```

```python
address.info()
```

```python
address.to_csv("data/cleaned/cleaned_addresses.csv", index=False)
print("Done")
```

## Road Segment

```python
road_segment = pd.read_csv("data/raw/City_of_Winnipeg_LRS_20260919.csv")
road_segment.head(5)
```

```python
road_segment.info()
```

```python
road_segment.shape
```

```python
road_segment.isna().sum()
```

```python
# Standardize column name
road_segment.columns = [col.strip().lower().replace(" ", "_") for col in road_segment.columns]

# Strip whitespaces
str_cols = ["street_name", "street_qualifier", "location"]
for col in str_cols:
    road_segment[col] = road_segment[col].astype(str).str.strip()

road_segment.head(5)
```

```python
road_segment.info()
```

```python
## Remove "," in begin measure, end measure
road_segment["begin_measure"] = (
    road_segment["begin_measure"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)
road_segment["end_measure"] = (
    road_segment["end_measure"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

## Convert to float
road_segment["begin_measure"] = pd.to_numeric(road_segment["begin_measure"], errors="coerce")
road_segment["end_measure"] = pd.to_numeric(road_segment["end_measure"], errors="coerce")

road_segment.isna().sum()
```

```python
road_segment.sample(5)
```

```python
road_segment.info()
```

```python
road_segment.duplicated().sum()
```

```python
road_segment.to_csv("data/cleaned/cleaned_road_segment.csv", index=False)
print("Done")
```

## Road Block

```python
road_block = pd.read_csv("data/raw/LRS_Block_Segments_20260919.csv")
road_block.head(5)
```

```python
# Drop unuseful cols
cols_to_drop = [
    "ID",
    "Block ID",
    "Street Qualfiier Secondary",
    "Address From Left",
    "Address To Left",
    "Address From Right",
    "Address To Right",
]

road_block = road_block.drop(columns=cols_to_drop, errors="ignore")
print(road_block.shape)
road_block.sample(5)
```

```python
road_block.isna().sum()
```

```python
road_block.info()
```

```python
# Standardize column names
road_block.columns = [col.strip().lower().replace(" ", "_") for col in road_block.columns]

# Strip whitespaces
str_cols = ["street_name", "street_qualifier", "street_class", "street_type_secondary", "location"]
for col in str_cols:
    road_block[col] = road_block[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

# Clean comma and convert measures to float
for measure_col in ["begin_measure", "end_measure"]:
    road_block[measure_col] = (
        road_block[measure_col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    road_block[measure_col] = pd.to_numeric(road_block[measure_col], errors="coerce")

# Cast lanes and oneway to int
road_block["number_of_lanes"] = road_block["number_of_lanes"].astype("Int64")
road_block["oneway"] = road_block["oneway"].astype("Int64")

road_block.info()
```

```python
road_block.sample(5)
```

```python
road_block.duplicated().sum()
```

```python
road_block.to_csv("data/cleaned/cleaned_road_block.csv", index=False)
print("Done")
```
