```python
import sqlite3
import pandas as pd
```

```python
# Connect to SQLite database
conn = sqlite3.connect("project.db")
cursor = conn.cursor()
```

```python
# Mapping of cleaned CSV files to target database table names
tables = {
    "data/cleaned/cleaned_addresses.csv": "addresses",
    "data/cleaned/cleaned_road_segment.csv": "road_segments",
    "data/cleaned/cleaned_road_block.csv": "road_blocks",
    "data/cleaned/cleaned_neighborhood.csv": "neighbourhoods",
    "data/cleaned/cleaned_plow_zone.csv": "plow_zones",
    "data/cleaned/cleaned_school_division.csv": "school_divisions",
    "data/cleaned/cleaned_speed_limit.csv": "speed_limits",
    "data/cleaned/cleaned_school_zone.csv": "school_zone",
}

# Load each cleaned CSV and write directly to SQLite
for csv_path, table_name in tables.items():
    df = pd.read_csv(csv_path)
    df.to_sql(table_name, conn, if_exists="replace", index=False)
    print(f"Table '{table_name}': loaded {len(df):,} rows.")
```

```python
conn.close()
```
